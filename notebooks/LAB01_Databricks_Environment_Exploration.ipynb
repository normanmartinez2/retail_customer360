```python
# LAB 01 - Create Retail Customer 360 catalog

spark.sql("""
    CREATE CATALOG IF NOT EXISTS retail_customer360
    COMMENT 'Retail Customer 360 - Data Engineering and Analytics Lab'
""")

print("Catalog created successfully!")
```

```python
# LAB 01 - Create Medallion Architecture schemas

schemas = {
    "bronze": "Raw ingested data from source systems",
    "silver": "Cleansed, validated and conformed data",
    "gold": "Business-ready datasets and analytics"
}

for schema_name, description in schemas.items():

    spark.sql(f"""
        CREATE SCHEMA IF NOT EXISTS
        retail_customer360.{schema_name}
        COMMENT '{description}'
    """)

    print(f"Schema created: {schema_name}")
```

```python
# Set current catalog and schema

spark.sql("USE CATALOG retail_customer360")
spark.sql("USE SCHEMA bronze")

# Verify configuration

display(
    spark.sql("""
        SELECT
            current_catalog() AS catalog,
            current_schema() AS schema
    """)
)
```
